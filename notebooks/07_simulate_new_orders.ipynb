```python
import pandas as pd
import numpy as np
import uuid
from sqlalchemy import create_engine

engine = create_engine('postgresql://postgres:Iftahyasmsm2003@localhost:5432/ecommerce_analytics')
```

```python
states = pd.read_sql("SELECT DISTINCT customer_state FROM customers", engine)['customer_state'].tolist()
categories = pd.read_sql("SELECT DISTINCT product_category_name_english FROM products WHERE product_category_name_english IS NOT NULL", engine)['product_category_name_english'].tolist()
```

```python
np.random.seed(None)  # different fake orders each run

n_fake = 5
fake_orders = pd.DataFrame({
    'order_id': [str(uuid.uuid4()) for _ in range(n_fake)],
    'customer_state': np.random.choice(states, n_fake),
    'product_category_name_english': np.random.choice(categories, n_fake),
    'order_value': np.round(np.random.uniform(30, 500, n_fake), 2),
    'num_items': np.random.randint(1, 4, n_fake),
    'freight_value': np.round(np.random.uniform(10, 80, n_fake), 2),
    'order_month': np.random.randint(1, 13, n_fake)
})

fake_orders
```

```python
fake_orders.to_sql('new_orders', engine, if_exists='append', index=False)
print("Inserted", len(fake_orders), "new orders.")
```
